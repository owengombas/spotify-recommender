```python
df_playlists = pd.DataFrame()
for u in users_list.USERS:
    df_playlist = u.get_playlists()
    df_playlist.to_json(os.path.join(data_path.DATA_PATH, f"{u.username}_playlists.json"), orient="records")
    df_playlists = pd.concat([df_playlists, df_playlist])
df_playlists
```

```python
all_user_playlists_tracks = pd.DataFrame()
for user in users_list.USERS:
    df_playlists_tracks = pd.DataFrame()
    for playlist_id in df_playlists[df_playlists["username"] == user.username]["id"]:
        df_current_playlist_tracks = user.get_playlist_tracks(playlist_id)
        df_playlists_tracks = pd.concat([df_playlists_tracks, df_current_playlist_tracks])
    df_playlists_tracks.to_json(os.path.join(data_path.DATA_PATH, f"{user.username}_playlists_tracks.json"), orient="records")
    all_user_playlists_tracks = pd.concat([all_user_playlists_tracks, df_playlists_tracks])
all_user_playlists_tracks
```
