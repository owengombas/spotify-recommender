```python
import pandas as pd
from lib import spoti
import os
```

```python
DATA_PATH = os.path.join("data")
```

```python
owen = spoti.SpotiUser("owen")
brenda = spoti.SpotiUser("brenda")
arthur = spoti.SpotiUser("arthur")
michelle = spoti.SpotiUser("michelle")
paul = spoti.SpotiUser("paul")
```

```python
USERS = [owen, brenda, arthur, michelle]
USERS = [paul]
```

```python
[u.get_auth_url() for u in USERS]
```

```python
[u.log_user() for u in USERS]
```

```python
USERS_DF = []
for user in USERS:
    user_top_tracks_medium_term = user.top_tracks(limit=500, offset=0, time_range="medium_term")
    user_top_tracks_short_term = user.top_tracks(limit=500, offset=0, time_range="short_term")
    user_top_tracks_long_term = user.top_tracks(limit=500, offset=0, time_range="long_term")
    user_top_tracks_medium_term["time_range"] = "medium_term"
    user_top_tracks_short_term["time_range"] = "short_term"
    user_top_tracks_long_term["time_range"] = "long_term"
    user_top_tracks: pd.DataFrame = pd.concat([user_top_tracks_medium_term, user_top_tracks_short_term, user_top_tracks_long_term])
    user_top_tracks.to_json(os.path.join(DATA_PATH, f"{user.username}_top_tracks.json"), orient="records")
```
